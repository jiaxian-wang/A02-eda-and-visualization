```python
# Your Phase 1 solution to the problem goes here.
#Q2.1
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('./A02-eda-and-visualization/data/ForeignGifts_edu.csv')
#Q2.2
df["Foreign Gift Amount"].hist(bins=20, grid=False)
plt.xlabel("Foreign Gift Amount")
plt.ylabel("Count")
df["Foreign Gift Amount"].describe()
```

The distribution of Foreign Gift Amount is strongly right-skewed. Most gifts are relatively small, but there are a few very large gifts. The mean is about 588,000, while the median is only about 95,000. The values range from about -538,000 to 100 million, and the very large maximum explains the long right tail on the graph.

```python
#Q2.3
counts = df["Gift Type"].value_counts()
counts /counts.sum()
```

About 61.21% of the gifts are contracts, 38.75% are monetary gifts, and only 0.04% are real estate.

```python
#Q2.4
import numpy as np
import seaborn as sns
df["log_gift"] = np.log(df["Foreign Gift Amount"])
sns.kdeplot(data=df, x="log_gift")
plt.xlabel("Log Foreign Gift Amount")
```

```python
sns.kdeplot(data=df, x="log_gift", hue="Gift Type")
plt.xlabel("Log Foreign Gift Amount")
```

After taking the log, the distribution is less right-skewed. Contracts are more concentrated around higher log gift amounts, while monetary gifts are more spread out. Real estate gifts are very rare.
